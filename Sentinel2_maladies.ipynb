```python
!pip install -q imagehash
import glob, os, imagehash, numpy as np
from PIL import Image

chemins = sorted(p for p in glob.glob("/content/bananalsd/**/*", recursive=True)
                 if p.lower().endswith((".jpg", ".jpeg", ".png")))
classes = [os.path.basename(os.path.dirname(p)) for p in chemins]
print(len(chemins), "images")

H = np.array([imagehash.phash(Image.open(p)).hash.flatten() for p in chemins]).astype(np.int32)
D = H @ (1 - H).T + (1 - H) @ H.T                      # distance de Hamming entre toutes les paires
iu = np.triu_indices(len(H), k=1)
paires = [(i, j, int(D[i, j])) for i, j in zip(*iu) if D[i, j] <= 5]

print("Paires très proches (distance <= 5) :", len(paires))
print("  dont exactement identiques :", sum(1 for _, _, d in paires if d == 0))
print("  entre classes différentes :", sum(1 for i, j, _ in paires if classes[i] != classes[j]))
for i, j, d in sorted(paires, key=lambda t: t[2])[:8]:
    print(d, "|", classes[i], "|", os.path.basename(chemins[i]), "<->", classes[j], "|", os.path.basename(chemins[j]))
```

```python
import pandas as pd
from scipy.sparse import csr_matrix
from scipy.sparse.csgraph import connected_components

n_groupes, groupe = connected_components(csr_matrix(D <= 5), directed=False)
df_g = pd.DataFrame({"path": chemins, "classe": classes, "groupe": groupe})
df_g["copie"] = df_g["path"].str.contains("Copie", case=False)

tailles = df_g.groupby("groupe").size()
mixtes = df_g.groupby("groupe")["classe"].nunique()
ambigus = set(mixtes[mixtes > 1].index)

print("Fichiers :", len(df_g), "| images distinctes (groupes) :", n_groupes)
print("Fichiers dont le nom contient 'Copie' :", int(df_g["copie"].sum()))
print("Groupes de taille 1 / 2 / 3 ou plus :", int((tailles == 1).sum()), int((tailles == 2).sum()), int((tailles >= 3).sum()))
print("Groupes qui mélangent plusieurs classes (écartés) :", len(ambigus))

dedup = df_g.sort_values("copie").drop_duplicates("groupe", keep="first")
dedup = dedup[~dedup["groupe"].isin(ambigus)]
print("\nAprès nettoyage :", len(dedup), "images")
print(dedup["classe"].value_counts().to_string())

dedup.to_csv("/content/drive/MyDrive/fieldscan/data/bananalsd_originales/bananalsd_originales_dedup.csv", index=False)
print("Liste sauvegardée")
```

```python
import os
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Chemins des dossiers et du CSV
csv_path = '/content/drive/MyDrive/fieldscan/data/bananalsd_originales/bananalsd_originales_dedup.csv'
base_output_dir = '/content/drive/MyDrive/fieldscan/data/bananalsd_dataset_split'

# 2. Charger le CSV nettoyé
df = pd.read_csv(csv_path)
print(f"Chargement de {len(df)} images depuis le CSV.")

# 3. Créer l'arborescence train/val pour chaque classe
classes = df['classe'].unique()
for split in ['train', 'val']:
    for c in classes:
        os.makedirs(os.path.join(base_output_dir, split, c), exist_ok=True)

# 4. Répartir et copier les images (80% train, 20% val)
train_count = 0
val_count = 0

for c in classes:
    # Filtrer les images de la classe en cours
    df_class = df[df['classe'] == c]

    # Split train/val (80% / 20%)
    train_df, val_df = train_test_split(df_class, test_size=0.2, random_state=42)

    # Copie vers train
    for _, row in train_df.iterrows():
        src_path = row['path']
        if os.path.exists(src_path):
            filename = os.path.basename(src_path)
            dst_path = os.path.join(base_output_dir, 'train', c, filename)
            shutil.copy(src_path, dst_path)
            train_count += 1

    # Copie vers val
    for _, row in val_df.iterrows():
        src_path = row['path']
        if os.path.exists(src_path):
            filename = os.path.basename(src_path)
            dst_path = os.path.join(base_output_dir, 'val', c, filename)
            shutil.copy(src_path, dst_path)
            val_count += 1

print(f"\n Répartition terminée avec succès !")
print(f" Images d'entraînement copiées : {train_count}")
print(f" Images de validation copiées : {val_count}")
print(f" Dossier prêt à l'emploi : {base_output_dir}")
```

```python
from collections import Counter
from ultralytics import YOLO

def run_banana_inference():
    # 1. Charger les poids de ton entraînement (copie sur le Drive)
    model_path = '/content/drive/MyDrive/fieldscan/models/yolov8n_cls_banana_lsd_run/weights/best.pt'
    model = YOLO(model_path)
    print(f"Modèle chargé avec succès depuis : {model_path}")

    # 2. Le dossier d'images à tester (feuilles de Sigatoka de la validation)
    source_path = '/content/drive/MyDrive/fieldscan/data/bananalsd_dataset_split/val/sigatoka/'

    # 3. Prédiction : une classe par image (en classification, le paramètre conf n'est pas utile)
    results = model.predict(source=source_path, imgsz=224, save=True)

    # 4. Résumé : que répond le modèle sur des feuilles qui sont toutes de la Sigatoka ?
    reponses = Counter(r.names[r.probs.top1] for r in results)
    total = sum(reponses.values())
    print("\nRéponses du modèle sur", total, "images de Sigatoka :")
    for classe, n in reponses.most_common():
        print(f"  {classe}: {n} ({round(100 * n / total)} %)")
    return results

if __name__ == '__main__':
    run_banana_inference()
```

```python
import os, glob, zipfile, numpy as np, pandas as pd
from ultralytics import YOLO

DRIVE = "/content/drive/MyDrive/fieldscan"
modele = YOLO(f"{DRIVE}/models/yolov8n_cls_banana_lsd_run/weights/best.pt")
noms = modele.names
print("Classes du modèle :", noms)

def predire(chemins):
    sorties = []
    for r in modele.predict(source=chemins, imgsz=224, stream=True, verbose=False):
        sorties.append(r.probs.data.cpu().numpy())
    return np.array(sorties)

# 1. Toute la validation du Bangladesh
racine = f"{DRIVE}/data/bananalsd_dataset_split/val"
chem_v, y_v = [], []
for c in sorted(os.listdir(racine)):
    for p in glob.glob(f"{racine}/{c}/*"):
        chem_v.append(p); y_v.append(c)
P_v = predire(chem_v)
pred_v = np.array([noms[i] for i in P_v.argmax(1)])
print("\\nVALIDATION BANGLADESH :", len(chem_v), "images")
print(pd.crosstab(pd.Series(np.array(y_v), name="réel"), pd.Series(pred_v, name="réponse")))
tau = float(np.quantile(P_v.max(1), 0.05))      # seuil du « je ne sais pas » fixé sur le Bangladesh
print("Seuil de confiance (95 % des images de validation acceptées) :", round(tau, 3))

# 2. Tanzanie (autre pays)
def ouvrir(zip_path, dest):
    if not os.path.exists(dest):
        with zipfile.ZipFile(zip_path) as z:
            z.extractall(dest)
ouvrir(f"{DRIVE}/data/tanzanie_test/tanzanie_512.zip", "/content/tanzanie")

def liste(nom_dossier, n=1500):
    p = sorted(x for x in glob.glob(f"/content/tanzanie/**/{nom_dossier}/*", recursive=True)
               if x.lower().endswith((".jpg", ".jpeg", ".png")))
    rng = np.random.default_rng(0)
    return [p[i] for i in rng.permutation(len(p))[:n]]

sains, malades = liste("HEALTHY"), liste("BLACK_SIGATOKA")
chem_t = sains + malades
y_t = np.array(["healthy"] * len(sains) + ["sigatoka"] * len(malades))
P_t = predire(chem_t)
pred_t = np.array([noms[i] for i in P_t.argmax(1)])
conf_t = P_t.max(1)

print("\\nTANZANIE :", len(chem_t), "images")
print(pd.crosstab(pd.Series(y_t, name="réel"), pd.Series(pred_t, name="réponse")))
sain, mal = y_t == "healthy", y_t == "sigatoka"
acc = conf_t >= tau
print("Feuilles saines classées malades (fausses alertes) :", round(float(np.mean(pred_t[sain] != "healthy")), 3))
print("Sigatoka reconnues comme Sigatoka :", round(float(np.mean(pred_t[mal] == "sigatoka")), 3))
print("Sigatoka classées saines (maladies ratées) :", round(float(np.mean(pred_t[mal] == "healthy")), 3))
print("Sigatoka classées autre maladie :", round(float(np.mean(~np.isin(pred_t[mal], ["sigatoka", "healthy"]))), 3))
print("Avec le seuil de rejet : couverture", round(float(acc.mean()), 3),
      "| fausses alertes", round(float(np.mean((pred_t[sain] != "healthy") & acc[sain])), 3))
print("Confiance médiane en Tanzanie :", round(float(np.median(conf_t)), 3))
```

```python
import os, glob, zipfile, time
import numpy as np, pandas as pd
from google.colab import drive
if not os.path.ismount("/content/drive"):
    drive.mount("/content/drive")
from ultralytics import YOLO

DRIVE = "/content/drive/MyDrive/fieldscan"
modele = YOLO(f"{DRIVE}/models/yolov8n_cls_banana_lsd_run/weights/best.pt")
noms = modele.names
print("Classes du modèle :", noms)

def predire(chemins):
    sorties = []
    for i, r in enumerate(modele.predict(source=chemins, imgsz=224, stream=True, verbose=False)):
        sorties.append(r.probs.data.cpu().numpy())
        if (i + 1) % 500 == 0:
            print(f"   {i + 1} / {len(chemins)} images traitées", flush=True)
    return np.array(sorties)

# 1. Validation du Bangladesh (toutes les classes)
racine = f"{DRIVE}/data/bananalsd_dataset_split/val"
chem_v, y_v = [], []
for c in sorted(os.listdir(racine)):
    for p in glob.glob(f"{racine}/{c}/*"):
        chem_v.append(p); y_v.append(c)
P_v = predire(chem_v)
pred_v = np.array([noms[i] for i in P_v.argmax(1)])
print("\nVALIDATION BANGLADESH :", len(chem_v), "images")
print(pd.crosstab(pd.Series(np.array(y_v), name="réel"), pd.Series(pred_v, name="réponse")))
tau = float(np.quantile(P_v.max(1), 0.05))
print("Seuil de confiance (95 % des images de validation acceptées) :", round(tau, 3))

# 2. Tanzanie (autre pays)
def ouvrir(zip_path, dest):
    if not os.path.exists(dest):
        print("Décompression de", os.path.basename(zip_path), "...", flush=True)
        with zipfile.ZipFile(zip_path) as z:
            z.extractall(dest)
ouvrir(f"{DRIVE}/data/tanzanie_test/tanzanie_512.zip", "/content/tanzanie")

def liste(nom_dossier, n=1000):
    p = sorted(x for x in glob.glob(f"/content/tanzanie/**/{nom_dossier}/*", recursive=True)
               if x.lower().endswith((".jpg", ".jpeg", ".png")))
    rng = np.random.default_rng(0)
    return [p[i] for i in rng.permutation(len(p))[:n]]

sains, malades = liste("HEALTHY"), liste("BLACK_SIGATOKA")
print("\nTanzanie :", len(sains), "saines et", len(malades), "Sigatoka (échantillon)", flush=True)
chem_t = sains + malades
y_t = np.array(["healthy"] * len(sains) + ["sigatoka"] * len(malades))
P_t = predire(chem_t)
pred_t = np.array([noms[i] for i in P_t.argmax(1)])
conf_t = P_t.max(1)

print("\nTANZANIE :", len(chem_t), "images")
print(pd.crosstab(pd.Series(y_t, name="réel"), pd.Series(pred_t, name="réponse")))
sain, mal = y_t == "healthy", y_t == "sigatoka"
acc = conf_t >= tau
print("Feuilles saines classées malades (fausses alertes) :", round(float(np.mean(pred_t[sain] != "healthy")), 3))
print("Sigatoka reconnues comme Sigatoka :", round(float(np.mean(pred_t[mal] == "sigatoka")), 3))
print("Sigatoka classées saines (maladies ratées) :", round(float(np.mean(pred_t[mal] == "healthy")), 3))
print("Sigatoka classées autre maladie :", round(float(np.mean(~np.isin(pred_t[mal], ["sigatoka", "healthy"]))), 3))
print("Avec le seuil de rejet : couverture", round(float(acc.mean()), 3),
      "| fausses alertes", round(float(np.mean((pred_t[sain] != "healthy") & acc[sain])), 3))
print("Confiance médiane en Tanzanie :", round(float(np.median(conf_t)), 3),
      "| médiane quand le modèle se trompe :", round(float(np.median(conf_t[pred_t != y_t])), 3) if np.any(pred_t != y_t) else "aucune erreur")

pd.DataFrame({"fichier": [os.path.basename(p) for p in chem_t], "reel": y_t, "reponse": pred_t,
              "confiance": conf_t.round(4)}).to_csv(f"{DRIVE}/outputs/yolo_tanzanie_par_image.csv", index=False)
print("\nRésultats sauvegardés dans outputs/yolo_tanzanie_par_image.csv")
```

```python
import os, time, zipfile, warnings
import numpy as np, rasterio
from scipy import ndimage as ndi
from skimage.feature import peak_local_max
from skimage.filters import threshold_otsu
from huggingface_hub import hf_hub_download
warnings.filterwarnings("ignore")

chemin = hf_hub_download("isp-uv-es/reforestree-taco", "reforestree.zip",
                         repo_type="dataset", local_dir="/content/reforestree")
dest = "/content/reforestree_extrait"
with zipfile.ZipFile(chemin) as z:
    z.extract("DATA/35/image.tif", dest)

gsd = 0.02

def exg(img):
    img = img.astype("float32")
    s = img.sum(axis=2) + 1e-6
    r, v, b = img[..., 0] / s, img[..., 1] / s, img[..., 2] / s
    return 2 * v - r - b

def compter(img, gsd_m, couronne_m, f=4):
    h, w, _ = img.shape
    h2, w2 = h // f, w // f
    p = img[:h2 * f, :w2 * f].reshape(h2, f, w2, f, 3).mean(axis=(1, 3))
    ind = exg(p)
    masque = ind > threshold_otsu(ind)
    diam = couronne_m / (gsd_m * f)
    lisse = ndi.gaussian_filter(masque.astype("float32"), sigma=diam / 6)
    pics = peak_local_max(lisse, min_distance=max(1, int(diam / 3)),
                          threshold_abs=0.3, labels=masque.astype(int))
    return (pics[:, ::-1] + 0.5) * f, masque

t0 = time.time()
with rasterio.open(f"{dest}/DATA/35/image.tif") as s:
    img = np.transpose(s.read([1, 2, 3]), (1, 2, 0))
t_lecture = time.time() - t0

surface_ha = img.shape[0] * img.shape[1] * gsd ** 2 / 10000
print("Processeurs disponibles :", os.cpu_count())
print("Tuile :", img.shape[:2], "| surface :", round(surface_ha, 2), "ha (si 2 cm par pixel)")

for f in (4, 8):
    t0 = time.time()
    for _ in range(3):
        compter(img, gsd, couronne_m=5.2, f=f)
    t_calcul = (time.time() - t0) / 3
    print(f"f={f} (résolution {int(gsd * 100 * f)} cm) : calcul {t_calcul:.1f} s | "
          f"lecture {t_lecture:.1f} s | total {t_calcul + t_lecture:.1f} s par tuile | "
          f"{(t_calcul + t_lecture) / surface_ha:.1f} s par hectare")
```
